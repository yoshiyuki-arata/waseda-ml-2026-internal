# オブジェクト指向プログラミングの入口

この Notebook では、`float`（浮動小数点）オブジェクトと `str`（文字列）オブジェクトを例にして、Python におけるオブジェクト指向プログラミングの基本を確認する。

ポイントは次の 3 つである。

- Python では、数値や文字列もオブジェクトとして扱われる
- オブジェクトは、データだけでなくメソッドや属性も持つ
- データと操作を一体として考えるのがオブジェクト指向の基本である


## 1. 文字列オブジェクトを見てみる

まずは `str` オブジェクトを使う。文字列は、オブジェクト指向の説明に最も向いている例のひとつである。


```python
s = "hello"

print(s)
print(type(s))
```

`"hello"` は単なる文字の並びではなく、`str` クラスから作られたオブジェクトである。

このオブジェクトは、自分に対して実行できるメソッドを持っている。


```python
print(s.upper())
print(s.count("l"))
print(s.replace("h", "H"))
```

ここでの `upper()` や `count()` は、文字列オブジェクトが持つ機能である。

たとえば `s.upper()` は、「文字列 `s` に対して大文字化という操作を行う」という意味になる。
これは、データと操作が一体になっていることを示している。


## 2. float オブジェクトも見てみる

次に `float` オブジェクトを見る。`3.14` のような数値も、Python ではオブジェクトである。


```python
x = 3.14
y = 3.0

print(type(x))
print(type(y))
```

`3.14` も `3.0` も `float` クラスのオブジェクトである。

`float` は `str` よりメソッドの数は少ないが、それでもオブジェクトとして振る舞う。


```python
print(x.is_integer())
print(y.is_integer())
```

`is_integer()` は、`float` オブジェクトが持つメソッドである。

つまり数値であっても、「その値に対して使える操作」がオブジェクトに結びついている。


```python
print(x.as_integer_ratio())
print(x.hex())
```

`as_integer_ratio()` や `hex()` も `float` オブジェクトのメソッドである。

このように、数値も「値」だけでなく「その値に対する操作」を持っている。


## 3. 属性にも注目する

オブジェクトはメソッドだけでなく、属性も持つことがある。


```python
print(x.real)
print(x.imag)
```

`real` や `imag` は属性であり、オブジェクトの性質を表している。

このことから、オブジェクトは「値」と「その値に関する情報や操作」をまとめて持っていると考えられる。


## 4. 演算子も実はメソッドである

オブジェクト指向を理解するうえで重要なのは、四則演算も内部的にはメソッドとして実装されている点である。


```python
a = 3.14
b = 2.0

print(a + b)
print(a.__add__(b))
```

`a + b` は、内部的には `a.__add__(b)` のような形で処理される。

つまり、ふつうの計算ですらオブジェクトのメソッド呼び出しとして実装されている。


```python
print(a.__sub__(b))
print(a.__mul__(b))
print(a.__truediv__(b))
```

この例から、数値オブジェクトも自分に対する振る舞いを持っていることが分かる。


## 5. `str` と `float` を並べて考える

`str` と `float` は見た目はかなり違うが、どちらもオブジェクトである。


```python
s = "hello"
x = 3.14

print(type(s))
print(type(x))
print(s.upper())
print(x.is_integer())
```

どちらもオブジェクトだが、持っているメソッドは同じではない。

- `str` は文字列操作のためのメソッドを持つ
- `float` は数値計算や数値表現に関するメソッドを持つ

ここから、「オブジェクトごとにできることが違う」という考え方が見えてくる。


## 6. オブジェクト指向プログラミングとしての見方

ここまでの例をオブジェクト指向という言葉でまとめると、次のようになる。

- オブジェクトはデータを持つ
- オブジェクトはそのデータに対する操作も持つ
- プログラムは、それぞれのオブジェクトに応じた操作を呼び出して動く

たとえば `"hello"` には `upper()` があり、`3.14` には `is_integer()` がある。
このように、データと操作をひとまとまりとして考えるのが、オブジェクト指向プログラミングの基本である。


## 7. まとめ

- Python では文字列も浮動小数点数もオブジェクトである
- オブジェクトは、型ごとに異なるメソッドや属性を持つ
- 演算子も内部的にはメソッドとして実装されている
- したがって Python では、身近なデータ型を通してオブジェクト指向の考え方を学べる

このあと `class` を使って自分でクラスを定義するときも、考え方の基本は同じである。
