## 4. 分析の再現可能性

分析の結果が良かった・悪かったの他に、データ分析で重要な点として、**再現可能性**がある。  
つまり、自分の行った分析が、他の第三者でも同じく再現できるかである。  
どんなに素晴らしい結果が得られたと主張しても、それを第三者が再現できなければ、それは信頼できる結果とは見なされない。

再現可能性を担保するために、いくつかの注意すべき点がある。

**乱数を固定する**

データを train/test data に分割する時や Random Forests モデルでは、乱数（ランダムな値）を発生させて、それを関数の中で用いている。  
コードを実行するたびに乱数が変わってしまうと、再現可能性の点から問題となる。   
そこで、乱数を作り出す元になる値（seed 値）を予め設定し、乱数を使う場合でも、実行のたびに同じ結果になるようにする。  
機械学習の多くの関数では seed 値を受け取れる設定になっているので、それを用いるとよい。  
seed 値を `42` に設定するのは単なる慣習だが、特に事情がなければ、`42` と設定する方がよい。



````python
import numpy as np
import random

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)
````


```python

X_train, X_test, y_train, y_test = train_test_split(

```


**データの入手プロセス**

難しい場合もあるが、データの入手のプロセス自体も、コードの実行によって自動的に行われるようにする方が、再現可能性の点では好ましい。
画面上でのクリック（GUIの操作）などは、後でもう一度再実行する時には忘れていることも多く、結果の再現ができなくなってしまうからである。
最も望ましいのは、データの取得から結果の出力まで一貫して jupyter notebook に記載し、全てのセルの実行（Run all）を押しさえすれば、同じ結果が必ず出力されるというものである。

例えば、UCI Machine Learning Repository の場合は、モジュールを使って notebook 上で直接ダウンロードできるので、再現可能性の点では好ましい。
（ただし、これは、データは必ず UCI Machine Learning Repository でなければならないという意味ではなく、データの取得に関しても十分に再現可能性に注意を払ってという意味である）
