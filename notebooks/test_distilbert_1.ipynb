```python
from transformers import pipeline

unmasker = pipeline('fill-mask', model='distilbert-base-uncased')
```

```python
unmasker("Hello I'm a [MASK] model.")
```

```python
unmasker("Ask not what your country can do for you, ask what you can do for your [MASK].")
```

```python
unmasker("I have a [MASK].")
```

```python
unmasker("I have a [MASK] that my four little children will one day live in a nation where they will not be judged by the color of their skin but by the content of their character.")
```
